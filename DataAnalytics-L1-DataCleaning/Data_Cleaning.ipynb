## Missing Value Handling

- `children`: Missing values were replaced with the median because it is a numerical column and the median is robust to outliers.
- `country`: Missing values were filled with the most frequent country (mode) because it is a categorical feature.
- `agent` and `company`: Missing values were replaced with `0` to indicate that no travel agent or company was associated with the booking.

```python
# Fill missing values in 'children' with the median
df['children'] = df['children'].fillna(df['children'].median())

# Fill missing values in 'country' with the mode
df['country'] = df['country'].fillna(df['country'].mode()[0])

# Fill missing values in 'agent' with 0
df['agent'] = df['agent'].fillna(0)

# Fill missing values in 'company' with 0
df['company'] = df['company'].fillna(0)
```

```python
df.isnull().sum()
```

## Missing Value Handling

The missing values were handled using different strategies based on the type and meaning of each column:

- `children`: Missing values were replaced with the median because it is a numerical column and the median is less affected by outliers.
- `country`: Missing values were filled with the most frequent value (mode) since it is a categorical feature.
- `agent`: Missing values were replaced with `0`, indicating that no travel agent was associated with the booking.
- `company`: Missing values were replaced with `0`, indicating that no company booking was associated with the reservation.

After applying these techniques, the dataset contains no missing values.

## Data Type Correction

The `reservation_status_date` column was converted from an object data type to `datetime` to enable efficient date-based analysis and ensure correct data representation.

```python
# Select numerical columns
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

for col in numerical_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower) | (df[col] > upper)).sum()

    print(f"{col}: {outliers} outliers")
```

## Outlier Detection

The Interquartile Range (IQR) method was used to identify outliers in all numerical columns.

Outliers were detected in features such as `lead_time`, `adr`, `booking_changes`, and `adults`.

After careful evaluation, these outliers were **retained** because they represent genuine hotel booking scenarios (e.g., long booking lead times, premium room prices, or large group bookings) rather than data entry errors.

Retaining these values preserves valuable business information and ensures that meaningful observations are not removed from the dataset.
